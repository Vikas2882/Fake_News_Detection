```python
import matplotlib.pyplot as plt
import numpy as np

# Data
data_labels = ['KNN', 'SVM', 'NB', 'DT', 'Bagging', 'Adaboost', 'BERT', 'Optimized BERT']
accuracy = [90.16, 96.73, 92.12, 89.92, 95.31, 95.32, 97.72, 99.88]
f_score = [89.78, 96.56, 91.85, 89.24, 95.00, 95.02, 97.79, 99.88]

x = np.arange(len(data_labels))  # Label locations
width = 0.4  # Bar width

# Plot
fig, ax = plt.subplots(figsize=(10, 6))
rects1 = ax.bar(x - width/2, accuracy, width, label='Accuracy', color='steelblue')
rects2 = ax.bar(x + width/2, f_score, width, label='F1-Score', color='salmon')

# Labels and Title
ax.set_xlabel("Methods")
ax.set_ylabel("Percentage")
ax.set_title("Comparison of Accuracy and F1-Score", pad=20)
ax.set_xticks(x)
ax.set_xticklabels(data_labels, rotation=90, ha="right")
# ax.legend()
ax.legend(bbox_to_anchor=(0.1, -0.15), loc='upper center', borderaxespad=0.)
ax.set_ylim(0, 102)
ax.yaxis.set_major_formatter(plt.FuncFormatter(lambda x, _: f'{x:.2f}'))

# Add grid lines
ax.yaxis.grid(True, linestyle='-', linewidth=0.6, color='grey')

# Show plot
plt.tight_layout()
plt.show()
```
