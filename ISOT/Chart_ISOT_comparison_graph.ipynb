```python
import matplotlib.pyplot as plt
import numpy as np

# Data
data_labels = ['LSTM', 'BiLSTM', 'CNN-BiLSTM', 'GPT-4', 'Gemini Pro 1.0', 'Mistral 7B Instruct', 'Gemma 7B', 'BERT', 'Optimized BERT']
accuracy = [96.97, 97.10, 97.22, 96.70, 96.60, 90.00, 93.33, 98.74, 99.98]
f_score = [96.97, 97.10, 97.22, 96.54, 96.54, 90.90, 90.90, 99.00, 99.98]

x = np.arange(len(data_labels))  # Label locations
width = 0.4  # Bar width

# Plot
fig, ax = plt.subplots(figsize=(10, 6))
rects1 = ax.bar(x - width/2, accuracy, width, label='Accuracy', color='steelblue')
rects2 = ax.bar(x + width/2, f_score, width, label='F1-Score', color='salmon')

# Labels and Title
ax.set_xlabel("Methods")
ax.set_ylabel("Percentage")
ax.set_title("Comparison of Accuracy and F1-Score", pad=20)
ax.set_xticks(x)
ax.set_xticklabels(data_labels, rotation=90, ha="right")
# ax.legend()
ax.legend(bbox_to_anchor=(0.1, -0.3), loc='upper center', borderaxespad=0.)
ax.set_ylim(0, 102)
ax.yaxis.set_major_formatter(plt.FuncFormatter(lambda x, _: f'{x:.2f}'))

# Add grid lines
ax.yaxis.grid(True, linestyle='-', linewidth=0.6, color='grey')

# Show plot
plt.tight_layout()
plt.show()
```
